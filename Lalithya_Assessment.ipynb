```python
import numpy as np
import pandas as pd
np.random.seed(42)
n = 200
df = pd.DataFrame({
"order_id": range(1001, 1001 + n),
"order_date": pd.date_range("2024-01-01", periods=n, freq="D"),
"customer_name": np.random.choice(
["Anita", "Ravi", "Meera", "Arjun", "Sneha", "Karthik", "Divya",
"Suresh"], n),
"city": np.random.choice(
["Kochi", "Chennai", "Bengaluru", "Hyderabad", "Mumbai"], n),
"category": np.random.choice(
["Electronics", "Clothing", "Grocery", "Books", "Furniture"], n),
"quantity": np.random.randint(1, 10, n),
"unit_price": np.random.randint(100, 5000, n),
"payment_mode": np.random.choice(
["UPI", "Card", "Cash", "NetBanking"], n),
})
df.to_csv("sales_data.csv", index=False)
print("File created with", len(df), "rows")
```

```python

def calculate_total(quantity, unit_price, discount_percent=0):
    total = quantity*unit_price
    discount = total*discount_percent / 100
    final_amount = total-discount
    return final_amount
print(calculate_total(3, 500))
print(calculate_total(2, 1000, 10))
```

```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("sales.db")
query = """
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    order_date TEXT,
    customer_name TEXT,
    quantity INTEGER,
    city TEXT,
    category TEXT,
    unit_price INTEGER,
    payment_mode TEXT
);
"""
conn.execute(query)
print("Table created successfully")
```

```python

sales["order_date"] = sales["order_date"].astype(str)
sales[
    ["order_id", "order_date", "customer_name", "quantity",
     "city", "category", "unit_price", "payment_mode"]
].to_sql(
    "orders",
    conn,
    if_exists="replace",
    index=False
)
print("Data inserted successfully")
```

```python
count_query = "SELECT COUNT(*)FROM orders"
result = pd.read_sql_query(count_query, conn)
print(result)
```

```python
query = "SELECT*FROM orders"
df = pd.read_sql_query(query, conn)
print(df)
```

```python
query1 = """
SELECT *
FROM orders
WHERE city = 'Mumbai';
"""
ord = pd.read_sql_query(query1, conn)
print(ord)
```

```python
query2 = """
SELECT category,
       SUM(quantity) AS total_quantity
FROM orders
GROUP BY category;
"""
quantity = pd.read_sql_query(query2, conn)
print(quantity)
```

```python
query3 = """
SELECT city,
       AVG(unit_price) AS average_price
FROM orders
GROUP BY city
HAVING AVG(unit_price) > 2500;
"""
avg_unitprice = pd.read_sql_query(query3, conn)
print(avg_unitprice)
```

```python
query4 = """
SELECT *,
       (quantity * unit_price) AS total_amount
FROM orders
ORDER BY total_amount DESC
LIMIT 5;
"""
most_exp = pd.read_sql_query(query4, conn)
print(most_exp)
```

**SECTION F -ETL PIPELINE**

```python
import pandas as pd
def extract(file_path):
    df = pd.read_csv(file_path)
    print("Number of rows loaded:", len(df))
    return df
```

```python
sales_data = extract("sales_data.csv")
```

```python
def transform(df):
    df["total_amount"] = df["quantity"] * df["unit_price"]
    df["order_size"] = df["total_amount"].apply(
        lambda x: "Large" if x >= 15000
        else "Medium" if x >= 5000
        else "Small"
    )
    df = df.drop_duplicates(subset="order_id")
    df = df[
        ["order_id", "order_date", "city",
         "category", "total_amount", "order_size"]
    ]
    return df
```

```python
import sqlite3
def load(df, db_name, csv_name):
    conn = sqlite3.connect(db_name)
    df.to_sql(
        "clean_orders",
        conn,
        if_exists="replace",
        index=False
    )
    df.to_csv(csv_name, index=False)
    print("Data saved successfully into SQLite database")
    print("Data saved successfully into CSV file")
    conn.close()
```

```python
df = extract("sales_data.csv")
```

```python
cleaned_df = transform(df)

print(cleaned_df.head())
```

```python
load(cleaned_df, "retail.db", "clean_orders.csv")
```

```python
conn = sqlite3.connect("retail.db")
query = """
SELECT order_size, COUNT(*) AS total_orders
FROM clean_orders
GROUP BY order_size;
"""
result = pd.read_sql_query(query, conn)
print(result)
conn.close()
```
